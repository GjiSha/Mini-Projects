```python
import pandas as pd
import numpy as np
import google_colab_selenium as gs
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
driver=gs.Chrome()
```

```python
driver.get('https://www.multibuzz.app/')
```

```python

try:
  room=WebDriverWait(driver,10).until(EC.presence_of_element_located((By.CSS_SELECTOR,"input#room")))
  room.clear()
  room.send_keys(Room_code)

  name=WebDriverWait(driver,10).until(EC.presence_of_element_located((By.CSS_SELECTOR,"input#name")))
  name.clear()
  name.send_keys(Team_name)

  submit_button = WebDriverWait(driver, 10).until(
         EC.presence_of_element_located((By.CSS_SELECTOR, "form.lobby-form button[type='submit']"))
     )
  submit_button.click()
except Exception as e:
  print(f"error occured: {e}")
finally:
  pass

```
